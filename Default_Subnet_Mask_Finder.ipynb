```python
first = int(input("Enter first octet: "))

if first <= 126:
    print("Class A")
    print("Default Mask: 255.0.0.0")

elif first <= 191:
    print("Class B")
    print("Default Mask: 255.255.0.0")

elif first <= 223:
    print("Class C")
    print("Default Mask: 255.255.255.0")

else:
    print("Special Class")
```
