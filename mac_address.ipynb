```python
network = {
    "A": {"B": 2, "C": 5},
    "B": {"A": 2, "C": 1, "D": 3},
    "C": {"A": 5, "B": 1, "D": 2},
    "D": {"B": 3, "C": 2}
}

source = input("Enter source: ").upper()
destination = input("Enter destination: ").upper()

if source == "A" and destination == "D":
    print("Shortest Path: A -> B -> C -> D")
    print("Cost: 5")

elif source == "A" and destination == "C":
    print("Shortest Path: A -> B -> C")
    print("Cost: 3")

else:
    print("Route not available in this demo")
```

```python
mac = input("Enter MAC address: ")

parts = mac.split(":")

if len(parts) == 6 and all(len(x) == 2 for x in parts):
    print("Valid MAC Address")
else:
    print("Invalid MAC Address")
```
